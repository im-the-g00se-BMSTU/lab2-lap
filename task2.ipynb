# Задание 2. Визуализация цен на товары

```python
import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv('wheat_coffee_corn_prizes.csv', parse_dates=['Date'])
```

## 1. Средние цены по годам

```python
annual_prices = data.groupby(data['Date'].dt.year)[['Wheat', 'Corn', 'Coffe']].mean()

fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(annual_prices.index, annual_prices['Wheat'], label='Пшеница')
ax.plot(annual_prices.index, annual_prices['Corn'], label='Кукуруза')
ax.plot(annual_prices.index, annual_prices['Coffe'], label='Кофе')
ax.set_title('Средние цены на пшеницу, кукурузу и кофе по годам')
ax.set_xlabel('Год')
ax.set_ylabel('Средняя цена')
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```

**Что показано:** средние за год цены пшеницы, кукурузы и кофе.

**Вывод:** пшеница и кукуруза достигли наибольших годовых средних цен в 2022 году: примерно 9,52 и 6,94. У кофе максимум был раньше - в 2011 году (около 2,53), а минимум - в 2002 году (около 0,54). Поэтому изменения цены кофе не всегда совпадают с изменениями цен двух зерновых культур.

## 2. Распределение цен на кофе

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(data['Coffe'], bins=30, edgecolor='white', label='Кофе')
ax.set_title('Распределение цен на кофе')
ax.set_xlabel('Цена кофе')
ax.set_ylabel('Количество наблюдений')
ax.legend()
plt.show()
```

**Что показано:** сколько наблюдений попадает в каждый диапазон цен на кофе.

**Вывод:** половина значений находится примерно между 0,96 и 1,51; медиана равна 1,25.

## 3. Связь цен на пшеницу и кукурузу

```python
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(data['Wheat'], data['Corn'], s=12, alpha=0.2,
           label='Пшеница и кукуруза')
ax.set_title('Связь цен на пшеницу и кукурузу')
ax.set_xlabel('Цена пшеницы')
ax.set_ylabel('Цена кукурузы')
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```

**Что показано:** каждая точка соответствует ценам пшеницы и кукурузы в одну дату.

**Вывод:** более высокой цене пшеницы обычно соответствует более высокая цена кукурузы.
